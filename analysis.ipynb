```python
# 데이터 전처리
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from lightgbm import LGBMClassifier

# Attr37 → 플래그 (공통)
df['no_LT_debt'] = df['Attr37'].isnull().astype(int)
df = df.drop(columns='Attr37')
X = df.drop(columns='class')
y = df['class']

# 상하위 1% 자르기: fit에서 경계를 배우고 transform에서 적용 (StandardScaler와 같은 구조)
class Winsorizer(BaseEstimator, TransformerMixin):
    def __init__(self, lower=0.01, upper=0.99):
        self.lower, self.upper = lower, upper
    def fit(self, X, y=None):
        X = pd.DataFrame(X)
        self.low_, self.high_ = X.quantile(self.lower), X.quantile(self.upper)
        return self
    def transform(self, X):
        return pd.DataFrame(X).clip(self.low_, self.high_, axis=1).values

models = {
    '로지스틱': make_pipeline(
        SimpleImputer(strategy='median'),   # 결측 → 중앙값
        Winsorizer(),                       # 극단값 자르기
        StandardScaler(),                   # 스케일링
        LogisticRegression(class_weight='balanced', max_iter=1000),
    ),
    '랜덤포레스트': make_pipeline(
        SimpleImputer(strategy='median'),
        RandomForestClassifier(n_estimators=300, class_weight='balanced',
                               random_state=42, n_jobs=-1),
    ),
    'LightGBM': LGBMClassifier(class_weight='balanced', random_state=42, verbose=-1),
}
```
